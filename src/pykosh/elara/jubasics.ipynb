# PyKosh Basics Showcase
This notebook demonstrates the basic Python bindings for the `kosh` Rust project, including mathematical operations, string handling, stateful classes, and hardware simulations.

```python
import pykosh
print("PyKosh module loaded successfully!")
```

## 1. Basic Functions
Showcasing simple pure Rust functions exposed to Python.

```python
print("add(3, 4) =", pykosh.add(3, 4))
print("greet('Elara') =", pykosh.greet("Elara"))
print("fibonacci(10) =", pykosh.fibonacci(10))
```

## 2. Stateful Classes
Showcasing a Rust `struct` exposed as a Python class.

```python
counter = pykosh.Counter(start=10)
print("Initial:", counter)
counter.increment(5)
print("After increment(5):", counter)
counter.reset()
print("After reset():", counter)
```

## 3. Rube Hardware Simulator
Showcasing the `add_via_rube` function, which calculates the sum of two integers by simulating a 32-bit hardware ripple-carry adder inside the `rube` engine.

```python
print("add_via_rube(10, 20) =", pykosh.add_via_rube(10, 20))
print("add_via_rube(12345, 54321) =", pykosh.add_via_rube(12345, 54321))
```
